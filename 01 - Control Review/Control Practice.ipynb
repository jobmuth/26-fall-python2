```python
def is_perfect(input_int):
    the_sum = 0
    for num in range(1 , input_int):
        if input_int % num == 0:
            the_sum = num + the_sum

    if the_sum == input_int:
        return True
    else:
        return False
```

```python
result = is_perfect(496)
print(result)
```

```python
result = is_perfect(100)
print(result)
```
