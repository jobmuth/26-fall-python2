### Practice

Create a function that takes two lists of integers as inputs (call them list1, list2, and list_sum). The function should return True if there is a solution to x + y = z, where x comes from list1, y comes from list2, and z comes from list_sum. Return False, otherwise. For example, list1 = [1, 2], list2 = [1, 3], and list_sum = [5, 6] should return true since you can find 2 + 3 = 5.

```python
def sum_check(list1, list2, list_sum):
    for num1 in list1:
        for num2 in list2:
            for sum in list_sum:
                if num1 + num2 == sum:
                    print(num1, num2, sum)
                    return True

    return False
```

```python
list1 = [1,2,3]
list2 = [4,5,6]
list_sum = [7,8,9]

sum_check(list1,list2,list_sum)
```

### Practice

Write a function that takes two equal-length lists of strings as input (call them key_list and val_list). The function should return a dictionary with the key-value pairs key_list[0] with val_list[0], key_list[1] with val_list[1], and so forth. For example, key_list = ['a', 'b'] and val_list = [1, 2] should return the dictionary {'a':1, 'b':2}.

```python
def create_dict(key_list, val_list):

    new_dict = dict()

    for loc in range(len(key_list)):
        Current_key = key_list[loc]
        Current_val = val_list[loc]
        new_dict[Current_key] = Current_val

    return new_dict
```

```python
create_dict(['a','b'],['x','y'])
```

### Practice

Write a function that takes two sets as input (call them set1 and set2). The function should return True is set1 and set2 have any elements in common. Otherwise, return False. (There are some clever set functions to handle this, but stick to using a for loop to scan over the first set and check whether that element belongs to the other set.)

```python
def has_intersection(set1, set2):

    for element in set1:
        for element in set2:
            return True

    return False
```

```python
has_intersection({1,2,3,4,5},{5,6,7,8})
```

### Practice

Modify your dictionary function to use the zip command: zip the two input lists at the start, then use a tuple to loop over both lists simultaneously and set up the key-value pairs.

```python
def create_dict_zip(key_list, val_list):

    new_dict = dict()

    zipped_pairs = zip(key_list, val_list)
    for (curr_key, curr_val) in zipped_pairs:
        new_dict[curr_key] = curr_val

    return new_dict
```
